```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "Department": ["HR", "Finance", "IT", "Sales"],
    "Absence": [5, 8, 16, 7]
})

plt.bar(df["Department"], df["Absence"],
        color=["green", "purple", "red", "purple"])
plt.title("Employee Absence")
plt.xlabel("Department")
plt.ylabel("Absence %")
plt.grid(True)
plt.legend(["Absence"])
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.bar(df["Department"], df["Absence"], color="navy")
plt.title("Employee Absence by Department")
plt.xlabel("Department")
plt.ylabel("Absence (%)")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.bar(df["Department"], df["Absence"], color="navy")
plt.title("Employee Absence by Department")
plt.xlabel("Department")
plt.ylabel("Absence (%)")
plt.grid(False)

ax = plt.gca()
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
plt.show()

```

```python
plt.figure(figsize=(8,5))
plt.bar(df["Department"], df["Absence"], color="navy")
plt.title("IT Has the Highest Absence — 16%")
plt.xlabel("Department")
plt.ylabel("Absence (%)")
plt.grid(False)

ax = plt.gca()
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

plt.annotate("Highest",
             xy=("IT", 25),
             xytext=("Finance", 20),
             arrowprops=dict(arrowstyle="->"))
plt.show()
```

### Visual Decluttering Principles Applied:
- Removed top and right spines to maximize data-ink ratio.
- Direct data labeling on bars to eliminate visual scanning across axes.
- Highlighting the primary insight (IT Dept: 16% absence) while muting secondary departments.

```python
import plotly.graph_objects as go
import numpy as np

departments = ["الموارد البشرية", "المالية", "تقنية المعلومات", "المبيعات"]
absence_rates = df["Absence"].values
avg_val = np.mean(absence_rates)

max_idx = np.argmax(absence_rates)
colors = ["#cbd5e1"] * len(absence_rates)
colors[max_idx] = "#0f766e"

fig = go.Figure()

fig.add_trace(go.Bar(
    x=departments,
    y=absence_rates,
    marker_color=colors,
    text=[f"<b>{val}%</b>" if i == max_idx else f"{val}%" for i, val in enumerate(absence_rates)],
    textposition="outside",
    hovertemplate="<b>القسم:</b> %{x}<br><b>نسبة الغياب:</b> %{y}%<extra></extra>",
    width=0.45
))

fig.add_hline(
    y=avg_val,
    line_dash="dash",
    line_color="#94a3b8",
    line_width=1.5,
    annotation_text=f"المتوسط العام ({avg_val:.1f}%)",
    annotation_position="top left",
    annotation_font=dict(color="#64748b", size=11)
)

fig.update_layout(
    title=dict(
        text="<b>تقنية المعلومات تسجل أعلى نسبة غياب متجاوزة المعدل العام</b>",
        font=dict(size=16, color="#0f172a"),
        x=0.5,
        xanchor="center"
    ),
    plot_bgcolor="white",
    paper_bgcolor="white",
    showlegend=False,
    margin=dict(t=80, b=50, l=40, r=60),
    xaxis=dict(
        title=dict(
            text="<b>الأقسام</b>",
            font=dict(size=12, color="#475569")
        ),
        showgrid=False,
        showline=True,
        linecolor="#94a3b8",
        linewidth=1.5,
        tickfont=dict(size=12, color="#334155", family="Arial Black")
    ),
    yaxis=dict(
        side="right",
        title=dict(
            text="<b>نسبة الغياب (%)</b>",
            font=dict(size=12, color="#475569")
        ),
        showgrid=False,
        showline=True,
        linecolor="#94a3b8",
        linewidth=1.5,
        showticklabels=True,
        dtick=4,
        ticksuffix="%",
        tickfont=dict(size=11, color="#334155", family="Arial Black")
    )
)

fig.show()
```
