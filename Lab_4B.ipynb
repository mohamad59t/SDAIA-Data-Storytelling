```python
measures = {
    "absence_rate": "نسبة الغياب",
    "absence_count": "عدد حالات الغياب"
}

print("اختر المؤشر:")
for key, name in measures.items():
    print(key, "→", name)
```

```python
import plotly.express as px

departments = ["HR", "Finance", "IT", "Sales"]
absence = [5, 8, 12, 7]

fig = px.bar(
    x=departments,
    y=absence,
    labels={"x": "Department", "y": "Absence %"},
    title="Absence by Department"
)

fig.show()
```
