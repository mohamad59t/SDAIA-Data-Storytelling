# Lab 6: Executive Delivery Clinic & Q&A Strategy

## Overview
This module transitions data storytelling from visual design into **Executive Delivery**.
It prepares the presenter to pitch recommendations within a strict 7-minute executive window and handle high-stakes committee Q&A.

### Key Focus Areas:
1. **7-Minute Presentation Time-Box:** Enforcing the BLUF narrative arc.
2. **Delivery Guardrails:** Orienting axes and highlighting primary marks without reading slides.
3. **Q&A Bridging Matrix:** Acknowledging concerns, answering crisply, and bridging back to the core decision.

```python
from dataclasses import dataclass

@dataclass
class StoryStep:
    arc_role: str
    message: str

# الفكرة الجوهرية
BIG_IDEA = "We Need To Invest More In The Lagging Regions."

# محطات القصة (عدّل النصوص حسب ما يناسب سيناريو الدورة)
STORYBOARD = [
    StoryStep("BLUF", "Prioritize investment in lagging regions to protect the 65% target."),
    StoryStep("situation", "Adoption is growing nationwide, but disparities are widening."),
    StoryStep("complication", "Lagging regions are pulling down overall adoption growth."),
    StoryStep("evidence", "lagging units show the highest performance gaps."),
    StoryStep("options", "Evaluate reallocation of Q4 resources versus status quo."),
    StoryStep("resolution", "Targeted regional support drives the highest return on investment."),
    StoryStep("ask", "Approve the reallocation of the next investment budget.")
]

TIME_BOX = {
    "BLUF": 30, "situation": 45,
    "complication": 60, "evidence": 75,
    "options": 90, "resolution": 60,
    "ask": 30
}   # seconds, sums to 6.5 min (under 7 min)

def build_notes(storyboard):
    total = 0
    print(f"OPEN (say the ask): {BIG_IDEA}\n")
    for s in storyboard:
        secs = TIME_BOX.get(s.arc_role, 45)
        total += secs
        print(f"[{total//60}:{total%60:02d}] {s.arc_role}")
        print(f"   Orient axes, then: {s.message}")
        print("   Point at one mark; do NOT read it.")
    assert total <= 7*60, f"Over by {total-420}s — cut."
    print("\nCLOSE (ask AGAIN): approve; 90-day check.")

build_notes(STORYBOARD)
```

```python
# story/qa_prep.py — acknowledge -> answer -> BRIDGE

QA = [
  {"q": "Aren't those regions just small? (hostile)",
   "answer": "Per-capita, not absolute (appx A2).",
   "bridge": "so targeted spend closes the real gap."},

  {"q": "Adoption looks high vs finance (data-wrong)",
   "answer": "Unique users, not a column avg (A5).",
   "bridge": "it holds at finance's lower figure."},

  {"q": "Why not spend it on the app? (off-topic)",
   "answer": "Channel infra is option 2; slower.",
   "bridge": "region-targeting is fastest to 65%."},
]

def drill(qa):
    for i, item in enumerate(qa, 1):
        print(f"Q{i}: {item['q']}")
        print(f"  A: {item['answer']}")
        print(f"  ->bridge: ...{item['bridge']}")

drill(QA)

```
