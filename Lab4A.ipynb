```python
import plotly.graph_objects as go

fig = go.Figure()

fig.add_trace(go.Indicator(
    mode="number",
    value=8,
    number={"suffix": "%"},
    title={"text": "Overall Absence"}
))

fig.show()

```

```python
import plotly.graph_objects as go

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
absence = [5, 6, 7, 8, 9, 8]

fig = go.Figure()

fig.add_trace(go.Scatter(
    x=months,
    y=absence,
    mode="lines+markers"
))

fig.update_layout(
    title="Monthly Absence Trend",
    xaxis_title="Month",
    yaxis_title="Absence (%)"
)

fig.show()
```

```python
import plotly.express as px

departments = ["HR", "Finance", "IT", "Sales"]
absence = [5, 8, 12, 7]

fig = px.bar(
    x=departments,
    y=absence,
    labels={"x": "Department", "y": "Absence (%)"},
    title="Absence by Department"
)

fig.show()
```

```python
employees = [100, 100, 100, 100]
absent = [5, 8, 12, 7]

total_absent = sum(absent)
total_employees = sum(employees)

rate = total_absent / total_employees * 100

print("Total absent:", total_absent)
print("Total employees:", total_employees)
print("Correct absence rate:", rate, "%")
```

```python
target = 5
current = 8

print("Current:", current, "%")
print("Target:", target, "%")

if current > target:
    print("Above target")
else:
    print("Within target")
```
